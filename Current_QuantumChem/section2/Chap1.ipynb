<h2 style="color:#9A11DA;"> What is Quantum Chemistry </h2> 

Quantum chemistry applies the principles of quantum mechanics to understand chemical systems at the molecular and atomic level. It is used to describe how electrons behave in atoms and molecules, and is essential in understanding chemical bonding, molecular properties, and reactions. To understand photosynthesis, or design new materials with specific properties, or understand how to design drugs, quantum chemistry is a crucial center. 

One of the central questions in quantum chemistry is solving Schr\"odinger's equation, but for a wavefunction that describes the electronic configuration of a molecule. Quantum chemistry calculations on classical computers typically scale as $\mathcal{O}(N^4) where $N$ is the number of orbitals. Number of orbitals can be thought of as a proxy for the size of the system, and the level of accuracy you get though we will learn more them later. 

So calculations start to get untractable very quickly!
